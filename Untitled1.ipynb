```python
training_data = [
    (["Sunny", "Warm", "Normal", "Strong", "Warm", "Same"], "Yes"),
    (["Sunny", "Warm", "High", "Strong", "Warm", "Same"], "Yes"),
    (["Rainy", "Cold", "High", "Strong", "Warm", "Change"], "No"),
    (["Sunny", "Warm", "High", "Strong", "Cool", "Change"], "Yes")
]
hypothesis = ["Ø"] * len(training_data[0][0])

for attributes, label in training_data:
    if label == "Yes":
        for i in range(len(hypothesis)):
            if hypothesis[i] == "Ø":
                hypothesis[i] = attributes[i]
            elif hypothesis[i] != attributes[i]:
                hypothesis[i] = "?"

print("Most Specific Hypothesis:")
print(hypothesis)
```

```python
# Candidate-Elimination Algorithm
training_data = [
    (["Sunny", "Warm", "Normal", "Strong", "Warm", "Same"], "Yes"),
    (["Sunny", "Warm", "High", "Strong", "Warm", "Same"], "Yes"),
    (["Rainy", "Cold", "High", "Strong", "Warm", "Change"], "No"),
    (["Sunny", "Warm", "High", "Strong", "Cool", "Change"], "Yes"),
]

num_attributes = len(training_data[0][0])
domains = [sorted({x[i] for x, _ in training_data}) for i in range(num_attributes)]
S = ["Ø"] * num_attributes
G = [["?"] * num_attributes]

def covers(hypothesis, example):
    return all(h == "?" or h == value for h, value in zip(hypothesis, example))

def more_general_or_equal(h1, h2):

    return all(a == "?" or a == b for a, b in zip(h1, h2))

for example, label in training_data:
    if label == "Yes":

        G = [g for g in G if covers(g, example)]

        for i, value in enumerate(example):
            if S[i] == "Ø":
                S[i] = value
            elif S[i] != value:
                S[i] = "?"

        G = [g for g in G if more_general_or_equal(g, S)]
    else:

        new_G = []
        for g in G:
            if covers(g, example):
                for i in range(num_attributes):
                    if g[i] == "?" and S[i] not in ("?", "Ø") and S[i] != example[i]:
                        h = g.copy()
                        h[i] = S[i]
                        if covers(h, example) is False and more_general_or_equal(h, S):
                            if h not in new_G:
                                new_G.append(h)
            else:
                new_G.append(g)
        G = new_G
    G = [g for g in G if not any(g != other and more_general_or_equal(g, other)
                                  for other in G)]

print("Specific Boundary (S):")
print(S)
print("General Boundary (G):")
for hypothesis in G:
    print(hypothesis)
```

```python

import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# Create dataset
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain',
                'Overcast', 'Sunny', 'Rain'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool',
                    'Cool', 'Mild', 'Mild'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes',
             'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

# Convert text values into numbers
le = LabelEncoder()
for col in df.columns:
    df[col] = le.fit_transform(df[col])

# Split input and output
X = df[['Outlook', 'Temperature']]
y = df['Play']

# Build ID3 decision tree
model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, y)

# Display decision tree
plot_tree(model, feature_names=['Outlook', 'Temperature'],
          class_names=['No', 'Yes'], filled=True)
plt.show()

# Classify a new sample: Sunny, Mild
new_sample = pd.DataFrame(
    [[le.fit_transform(['Sunny', 'Rain', 'Overcast']).tolist()[0],
      le.fit_transform(['Hot', 'Mild', 'Cool']).tolist()[1]]],
    columns=['Outlook', 'Temperature']
)

prediction = model.predict(new_sample)
print("Prediction:", "Yes" if prediction[0] == 1 else "No")
```
